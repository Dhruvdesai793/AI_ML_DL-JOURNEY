```python
# heatmap
# mostly used for corr

df4 = sns.load_dataset('titanic')
```

```python

import pandas as pd
df4
```

```python
#manual encoding

df4.sex = df4.sex.apply(lambda x: 1 if x == 'male' else 0)
df4.alive = df4.alive.apply(lambda x: 1 if x == 'yes' else 0)
df4.alone = df4.alone.apply(lambda x: 1 if x else 0)

df4['class'] = df4['class'].astype(pd.CategoricalDtype(categories=['First','Second','Third'], ordered=True)).cat.codes
df4 = df4.drop(['embarked', 'who', 'adult_male', 'deck', 'embark_town'], axis = 1)
```
