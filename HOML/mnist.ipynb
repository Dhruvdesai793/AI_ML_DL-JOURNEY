```python
import pandas as pd
import numpy as np
```

```python
from tensorflow.keras.datasets import mnist

(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
X_train
```

```python
import matplotlib.pyplot as plt

def plot_dig(img_data):
    image = img_data.reshape(28,28)
    plt.imshow(image, cmap='binary')
    plt.axis('off')

some_digit = X_train[0]
plot_dig(some_digit)
plt.show()
y_train.dtype
```

## Training a Binary Classifier

```python
y_train_5 = (y_train == 5)
y_test_5 = (y_test == 5)
y_test[2]
print(X_train.shape)
X_train = X_train.reshape(-1, 784)
X_train.shape
print(np.unique(y_train_5, return_counts=True))
```

```python
from sklearn.linear_model import SGDClassifier

sgd_clf = SGDClassifier(random_state = 42)
sgd_clf.fit(X_train, y_train_5)
```

```python
import seaborn as sns
X_test = X_test.reshape(-1, 784)
y_pred = sgd_clf.predict(X_test)

from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(
    
    y_pred,y_test_5
)
plt.show()
```

```python
from sklearn.metrics import PrecisionRecallDisplay

y_scores = sgd_clf.decision_function(X_test)

PrecisionRecallDisplay.from_predictions(
    y_test_5,
    y_scores
)
plt.show()
```

```python
from sklearn.model_selection import cross_val_score
cross_val_score(sgd_clf, X_train, y_train_5, cv=3, scoring='accuracy')
```

```python
from sklearn.dummy import DummyClassifier

d_clf = DummyClassifier()
d_clf.fit(X_train, y_train_5)
d_clf.predict(X_train)
```

```python
# d_clf score using cv
cross_val_score(d_clf, X_train, y_train_5, cv=3, scoring='accuracy')
```

#### sgd got 95 acc and dummy got 90 acc

```python
from sklearn.metrics import roc_curve
y_scores = sgd_clf.decision_function(X_test)
fpr, tpr, thresholds = roc_curve(y_test_5, y_scores)
```
