# Black friday dataset eda and feature engineering, cleaning and preparing the data for model training

## Problem Statement

A retail company “ABC Private Limited” wants to understand the customer purchase behaviour (specifically, purchase amount) against various products of different categories. They have shared purchase summary of various customers for selected high volume products from last month. The data set also contains customer demographics (age, gender, marital status, city_type, stay_in_current_city), product details (product_id and product category) and Total purchase_amount from last month.

Now, they want to build a model to predict the purchase amount of customer against various products which will help them to create personalized offer for customers against different products.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df_train = pd.read_csv('black_friday/train.csv')
df_test = pd.read_csv('black_friday/test.csv')
df_train
```

```python
df_test
```

## first combine both train and test dataset to perform data preprocessing on both of them simultaneously 

```python
#concat both train and test, append for df is depreciated in new pandas 2.0

combined = pd.concat([df_train,df_test])
combined
```

### Removing useless cols, here for purchase prediction, `User_ID` is useless. so we'll remove it.

```python
combined.drop(['User_ID'], axis=1, inplace=True)
```

```python
combined
```

### categorical feats: gender, age, city_category, 

```python
pd.get_dummies(combined['Gender'])
```

```python
combined['Gender'] = combined['Gender'].map({'F':0,'M':1})
```
