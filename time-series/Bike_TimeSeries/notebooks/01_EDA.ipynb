```python
df['datetime'] = (
    pd.to_datetime(df['dteday']) + pd.to_timedelta(df['hr'], unit = 'h')
)

df = df.set_index('datetime')
```

```python
df
```

```python
plt.figure(figsize = (16, 5))

df['cnt'].plot()

plt.title('Bike Rentals Over Time')
plt.ylabel('Bike Count')

plt.show()
```
