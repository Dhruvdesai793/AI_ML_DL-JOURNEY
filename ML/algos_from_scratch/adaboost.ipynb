```python
import pandas as pd
import numpy as np
from mlxtend.plotting import plot_decision_regions
```

```python
df = pd.DataFrame()
```

```python

df['X1'] = [1,2,3,4,5,6,6,7,9,9]
df['X2'] = [5,3,6,8,1,9,5,8,9,2]
df['label'] = [1,1,0,1,0,1,0,1,0,0]
```

```python
df
```

```python
import seaborn as sns
sns.scatterplot(x=df.X1 , y = df.X2, hue = df.label)
```

```python
# in start u assign equal weight to all rows

df['weights'] = 1/df.shape[0]
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt1 = DecisionTreeClassifier(max_depth = 1)
```

```python
X = df.iloc[:, 0:2].values
y = df.iloc[:, 2].values
```

```python
dt1.fit(X,y)
```

```python
from sklearn.tree import plot_tree
plot_tree(dt1)
```

```python
plot_decision_regions(X,y,clf = dt1, legend = 2)
# 3 mistakes as 3 blue points are in the wrong region.
```

```python
df['y_pred'] = dt1.predict(X)
```

```python
df # row 2, 6 and 8 have wrong predictions.
```

```python
def calculate_model_weight(error): # error = summation of weights of row misclassified.
    return 0.5*np.log((1-error)/(error))
```

```python
# calc model weight
alpha1 = calculate_model_weight(0.3)
alpha1
```

```python
def update_row_weights(row, alpha = 0.423):
    if row.label == row.y_pred:
        return row.weights * np.exp(-alpha)
    else:
        return row.weights * np.exp(alpha)
```

```python
df['update_weights'] = df.apply(update_row_weights, axis=1)
```

```python
df
```

```python
df.update_weights.sum()
```

```python
# it is .91 not 1, so we need to normalize it.
```

```python
df['normalized_weights'] = df.update_weights / df.update_weights.sum()
```

```python
df
```

```python
df.normalized_weights.sum()
```

```python


df['cumsum_upper'] = np.cumsum(df['normalized_weights'])
     

df['cumsum_lower'] = df['cumsum_upper'] - df['normalized_weights']
     

df[['X1','X2','label','weights','y_pred','update_weights','cumsum_lower','cumsum_upper']]
     
```

```python


def create_new_dataset(df):

  indices = []

  for i in range(df.shape[0]):
    a = np.random.random()
    for index,row in df.iterrows():
      if row['cumsum_upper'] > a and a > row['cumsum_lower']:
        indices.append(index)
  return indices
     

index_values = create_new_dataset(df)

index_values
     

[4, 2, 0, 8, 3, 9, 6, 8, 4, 0]


second_df = df.iloc[index_values,[0,1,2,3]]
     

second_df
     
```
